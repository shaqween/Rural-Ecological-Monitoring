# Эксперимент с классическими алгоритмами ML

Тип задачи — **мультиклассовая классификация** (3 класса: Healthy / Stressed / Critical).
Целевой признак — `crop_health`.

**Почему классификация:** таргет категориальный; задача — предсказать состояние культуры
по агрометеорологическим показателям (температура, влажность, осадки, влажность почвы,
pH, освещённость, удобрения).

**Метрики (обоснование):**
- **macro-F1** — основная. Классы несбалансированы (Healthy ≫ Stressed ≫ Critical),
  accuracy обманчива; macro-F1 усредняет F1 по классам без учёта размера и штрафует
  за плохое качество на редких классах.
- **balanced accuracy** — среднее recall по классам, устойчиво к дисбалансу.
- **accuracy** — для сравнения с baseline и общей картины.
- Confusion matrix — для анализа ошибок.

**Воспроизводимость:** `RANDOM_STATE = 42` везде (сплиты, модели, CV).

## Baseline-модели

Для сравнения используем две «наивные» стратегии `DummyClassifier`:
- **most_frequent** — всегда предсказывает самый частый класс (Healthy). Нижняя граница.
- **stratified** — случайное предсказание пропорционально частотам классов.

Эти модели не используют признаки, поэтому задают «планку», которую обязаны
побить все реальные алгоритмы.

```python
baseline_rows = []
for strat in ["most_frequent", "stratified"]:
    clf = DummyClassifier(strategy=strat, random_state=RANDOM_STATE)
    clf.fit(X_train, y_train)
    pred = clf.predict(X_val)
    baseline_rows.append({
        "baseline": strat,
        "val_acc": accuracy_score(y_val, pred),
        "val_bal_acc": balanced_accuracy_score(y_val, pred),
        "val_macro_f1": f1_score(y_val, pred, average="macro", zero_division=0),
    })

baseline_df = pd.DataFrame(baseline_rows)
print(baseline_df.to_string(index=False))
```

`most_frequent` даёт высокий accuracy (0.742 = доля `Healthy`), но низкий
macro-F1 (0.284): модель полностью игнорирует `Stressed` и `Critical`.
`stratified` проигрывает по accuracy (0.587), но выигрывает по macro-F1
(0.326) — случайные попадания в редкие классы дают ненулевой F1.

Это наглядно показывает, что **accuracy обманчива** при дисбалансе:
она награждает игнорирование редких классов. Для задачи важнее macro-F1
и balanced accuracy.

```python
models_and_grids = {
    "LogReg": (
        LogisticRegression(max_iter=2000, random_state=RANDOM_STATE),
        {"clf__C": [0.01, 0.1, 1.0, 10.0],
         "clf__penalty": ["l2"],
         "clf__solver": ["lbfgs"]}
    ),
    "DecisionTree": (
        DecisionTreeClassifier(random_state=RANDOM_STATE),
        {"clf__max_depth": [3, 5, 10, None],
         "clf__min_samples_leaf": [1, 5, 20]}
    ),
    "RandomForest": (
        RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
        {"clf__n_estimators": [100, 300],
         "clf__max_depth": [5, 10, None],
         "clf__min_samples_leaf": [1, 5]}
    ),
    "HistGB": (
        HistGradientBoostingClassifier(random_state=RANDOM_STATE),
        {"clf__learning_rate": [0.05, 0.1, 0.2],
         "clf__max_depth": [3, 6, None],
         "clf__max_iter": [100, 300]}
    ),
}

# вспомогательная функция разворачивания сетки
def expand_grid(grid):
    keys = list(grid)
    for vals in product(*grid.values()):
        yield dict(zip(keys, vals))

for name, (_, grid) in models_and_grids.items():
    n = len(list(expand_grid(grid)))
    print(f"{name}: {n} комбинаций гиперпараметров")
```

Выбраны 4 алгоритма из разных семейств — чтобы сравнить линейную модель,
дерево, bagging- и boosting-ансамбли:

| Алгоритм | Семейство | Что проверяем |
|---|---|---|
| **LogisticRegression** | линейный | сила регуляризации `C` |
| **DecisionTreeClassifier** | дерево | глубина, мин. листьев, критерий |
| **RandomForestClassifier** | bagging | глубина, число деревьев, мин. листьев |
| **HistGradientBoosting** | boosting | learning_rate, глубина, итерации |


Итого проверяем 4 + 12 + 12 + 18 = **46 комбинаций** гиперпараметров
для 4 алгоритмов. Сетки подобраны так, чтобы покрыть режимы недообучения,
оптимума и переобучения.

**Вывод:** `crop_health` в датасете сгенерирован по формуле от
`soil_moisture` и `temperature`. Простое правило
«`soil_moisture < 25` → Critical, `< 40` → Stressed, иначе Healthy»
даёт macro-F1 ≈ 0.85–0.90. Дерево глубины 3 восстанавливает эту формулу
и достигает macro-F1 ≈ 0.999.

**Следствия:**
1. Абсолютные метрики (≈1.0) **не отражают** реальную предсказательную
   силу на проде — они отражают лишь сложность генератора.
2. Сравнение алгоритмов **всё ещё осмысленно**: LogReg (линейная модель)
   покажет, нелинейна ли граница; CV mean±std покажет стабильность;
   feature importance подтвердит доминирование `soil_moisture` и `temperature`.
3. Для реальной постановки задачи нужны сенсорные данные с шумом,
   а не синтетический генератор.

Эксперимент продолжается с этой оговоркой.

```python
def plot_metric_vs_param(df, model_name, param_x, param_hue=None,
                         metric="val_macro_f1"):
    sub = df[df.model == model_name]
    if sub.empty or param_x not in sub.columns:
        print(f"[skip] {model_name}: нет {param_x}")
        return
    plt.figure(figsize=(7, 4))
    hue = param_hue if (param_hue and param_hue in sub.columns) else None
    sns.lineplot(data=sub, x=param_x, y=metric, hue=hue, marker="o")
    plt.title(f"{model_name}: {metric} vs {param_x}" +
              (f" (hue={param_hue})" if hue else ""))
    plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

plot_metric_vs_param(val_df, "LogReg", "C")
plot_metric_vs_param(val_df, "DecisionTree", "max_depth", "min_samples_leaf")
plot_metric_vs_param(val_df, "RandomForest", "max_depth", "n_estimators")
```

Графики подтверждают гипотезу: `crop_health` сгенерирован по детерминированной формуле от `soil_moisture` и `temperature`.

- **LogReg:** потолок ~0.84 — линейная модель не может восстановить нелинейные пороги (`<25`, `<40`). Оптимум `C ∈ [1, 8]`, далее плато/просадка.
- **DecisionTree:** уже при `max_depth=3` F1 ≈ 0.999 — дерево полностью восстанавливает исходную формулу. `min_samples_leaf=20` слегка снижает качество (~0.993), но картина та же.
- **RandomForest:** качество линейно растёт с `max_depth` (от ~0.86 при 5 до ~0.985+ при 10). `n_estimators` 100 vs 300 влияет только на малых глубинах.


**Итог:** одиночное дерево глубины 3 — оптимальное и интерпретируемое решение (F1 ≈ 0.999). Логистическая регрессия принципиально ограничена линейностью, а ансамбли требуют избыточной глубины (до 10) для сопоставимого результата.

```python
import pandas as pd
from sklearn.base import clone
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline

RANDOM_STATE = 42
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

scoring = {
    "macro_f1": "f1_macro",
    "balanced_acc": "balanced_accuracy",
    "acc": "accuracy",
}

# Исключаем HistGradientBoosting
EXCLUDE = {"HistGB", "HistGradientBoosting", "HistGradientBoostingClassifier"}
models_and_grids = {
    name: (est, grid)
    for name, (est, grid) in models_and_grids.items()
    if name not in EXCLUDE
}

rows = []
for name, (estimator, param_grid) in models_and_grids.items():
    pipe = Pipeline([
        ("prep", clone(preprocessor)),
        ("clf", clone(estimator)),
    ])

    search = GridSearchCV(
        estimator=pipe,
        param_grid=param_grid,
        cv=cv,
        scoring=scoring,
        refit=False,
        n_jobs=-1,
        verbose=1,
        return_train_score=False,
    )
    search.fit(X_train_full, y_train_full)

    res = pd.DataFrame(search.cv_results_)
    res["model"] = name
    rows.append(res)

all_res = pd.concat(rows, ignore_index=True)

param_cols = [c for c in all_res.columns if c.startswith("param_clf__")]
for c in param_cols:
    all_res[c.replace("param_clf__", "")] = all_res[c]
all_res = all_res.drop(columns=param_cols)

cv_df = (
    all_res[
        ["model"]
        + [c.replace("param_clf__", "") for c in param_cols]
        + [
            "mean_test_macro_f1", "std_test_macro_f1",
            "mean_test_balanced_acc", "std_test_balanced_acc",
            "mean_test_acc", "std_test_acc",
        ]
    ]
    .rename(columns={
        "mean_test_macro_f1": "cv_macro_f1_mean",
        "std_test_macro_f1": "cv_macro_f1_std",
        "mean_test_balanced_acc": "cv_bal_acc_mean",
        "std_test_balanced_acc": "cv_bal_acc_std",
        "mean_test_acc": "cv_acc_mean",
        "std_test_acc": "cv_acc_std",
    })
    .sort_values("cv_macro_f1_mean", ascending=False)
    .reset_index(drop=True)
)

print(cv_df.head(15).to_string(index=False))
```

Результаты подтверждают, что `crop_health` сгенерирован по детерминированной формуле от `soil_moisture` и `temperature`.

- **DecisionTree** — лучшая модель: уже при `max_depth=3` macro-F1 ≈ 0.9984, а при `max_depth=5` и `max_depth=10/None` — 0.9988–0.9989. Прирост от увеличения глубины минимален, значит дерево восстанавливает правило почти сразу.
- **RandomForest** — хуже одиночного дерева (0.9937 vs 0.9989) и требует `max_depth=None` + `n_estimators=300`, чтобы приблизиться. Ансамбль «размывает» жёсткие пороги формулы, поэтому проигрывает простому дереву.
- **LogReg** — в топ-15 не попал: линейная модель принципиально не может восстановить нелинейные пороги (`<25`, `<40`), её потолок ~0.84.

**Итог:** оптимальное решение — `DecisionTreeClassifier(max_depth=3, min_samples_leaf=1)` с CV macro-F1 ≈ 0.9984: максимально простое, интерпретируемое и практически не уступает более сложным конфигурациям. Ансамбли и линейные модели для данной задачи избыточны .

```python
def plot_cv_metric_vs_param(df, model_name, param_x,
                            mean_col="cv_macro_f1_mean",
                            std_col="cv_macro_f1_std"):
    sub = df[df.model == model_name].copy()
    if sub.empty or param_x not in sub.columns:
        return
    agg = (sub.groupby(param_x)
              .agg(mean=(mean_col, "mean"), std=(std_col, "mean"))
              .reset_index())
    plt.figure(figsize=(7, 4))
    plt.errorbar(agg[param_x].astype(str), agg["mean"], yerr=agg["std"],
                 marker="o", capsize=4)
    plt.title(f"{model_name} (CV): {mean_col} vs {param_x}")
    plt.xlabel(param_x); plt.ylabel(mean_col)
    plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

plot_cv_metric_vs_param(cv_df, "LogReg", "C")
plot_cv_metric_vs_param(cv_df, "DecisionTree", "max_depth")
plot_cv_metric_vs_param(cv_df, "RandomForest", "max_depth")
```


- **LogReg:** резкий рост до `C ≈ 0.1` (с ~0.68 до ~0.825), дальше плато. Потолок линейной модели ~0.83; при `C=0.01` огромный разброс (std ≈ 0.02) — модель недорегуляризована. Оптимум `C ∈ [0.1, 1.0]`.
- **DecisionTree:** глубина 3, 5 и 10 даёт одинаковое качество (~0.997), доверительные интервалы перекрываются, разница незначима. Формула восстанавливается уже на 3-м уровне → берём `max_depth=3` за простоту.
- **RandomForest:** значимый рост с 5 до 10 (0.954 → 0.987), интервалы не перекрываются. На глубине 5 высокий разброс (std ≈ 0.015). Но даже на 10 RF уступает одиночному дереву — усреднение «размывает» жёсткие пороги.

**Итог:** CV подтверждает hold-out, но error bars показывают, где разница значима. Оптимум — `DecisionTree(max_depth=3)`: macro-F1 ≈ 0.997, минимальный разброс, максимальная интерпретируемость.

```python
HISTGB_NAMES = {"HistGB", "HistGradientBoosting", "HistGradientBoostingClassifier"}

val_df = val_df[~val_df["model"].isin(HISTGB_NAMES)].copy()
cv_df  = cv_df[~cv_df["model"].isin(HISTGB_NAMES)].copy()

param_candidates = ["C", "penalty", "solver", "max_depth",
                    "min_samples_leaf", "n_estimators", "learning_rate"]

common_params = [
    c for c in param_candidates
    if c in val_df.columns and c in cv_df.columns
]

val_keys = ["model"] + common_params
cv_keys  = ["model"] + common_params

merged = pd.merge(
    val_df[val_keys + ["val_macro_f1", "val_bal_acc", "val_acc"]],
    cv_df[cv_keys + [
        "cv_macro_f1_mean", "cv_macro_f1_std",
        "cv_bal_acc_mean", "cv_bal_acc_std",
        "cv_acc_mean", "cv_acc_std",
    ]],
    on=val_keys,
    how="outer",
)

final_df = (
    merged
    .sort_values(["cv_macro_f1_mean", "val_macro_f1"], ascending=False)
    .reset_index(drop=True)
)

num_cols = [c for c in final_df.columns if c.startswith(("val_", "cv_"))]
final_df[num_cols] = final_df[num_cols].round(4)

print(final_df.to_string(index=False))
```

- **DecisionTree** — лучшая модель. `max_depth=10` и `None` дают val F1 = 1.0000 и CV F1 ≈ 0.9989 (std 0.0014). Глубина 3 даёт CV F1 ≈ 0.9984 — разница незначима (std больше разницы), значит формула восстанавливается уже на 3-м уровне. `min_samples_leaf=1` лучше, чем 5 и 20: сглаживание только вредит точной формуле.
- **RandomForest** — хуже дерева: максимум CV F1 ≈ 0.9937 (`n_estimators=300`, `max_depth=None`). На малой глубине (5) и `min_samples_leaf=5` качество проваливается до 0.93–0.96 с большим разбросом (std до 0.026). Ансамбль «размывает» жёсткие пороги.
- **LogReg** — принципиально ограничена: потолок CV F1 ≈ 0.83 при `C=1.0`. При `C=0.01` деградация до 0.68. Линейная модель не может восстановить нелинейные пороги (`<25`, `<40`).
- **Сходимость val и CV:** тренды совпадают (val чуть оптимистичнее — выбор по `X_val`), CV даёт устойчивые оценки со std. Разница между `max_depth=3` и `10` у дерева статистически незначима → выбираем `max_depth=3` за простоту и интерпретируемость.

```python
best_row = final_df.iloc[0]
best_model_name = best_row["model"]
print("Лучшая модель:", best_model_name)
print(best_row.to_string())

estimator, _ = models_and_grids[best_model_name]

param_keys = ["C", "penalty", "solver", "max_depth",
              "min_samples_leaf", "n_estimators", "learning_rate"]
best_params = {}
for k in param_keys:
    if k in best_row.index and pd.notna(best_row[k]):
        best_params[f"clf__{k}"] = best_row[k]

if "clf__max_depth" in best_params:
    v = best_params["clf__max_depth"]
    best_params["clf__max_depth"] = None if pd.isna(v) else int(v)
if "clf__n_estimators" in best_params:
    best_params["clf__n_estimators"] = int(best_params["clf__n_estimators"])
if "clf__min_samples_leaf" in best_params:
    best_params["clf__min_samples_leaf"] = int(best_params["clf__min_samples_leaf"])

print("Параметры:", best_params)

best_pipe = Pipeline([
    ("prep", clone(preprocessor)),
    ("clf", clone(estimator)),
])
best_pipe.set_params(**best_params)
best_pipe.fit(X_train_full, y_train_full)

y_pred = best_pipe.predict(X_test)

test_metrics = {
    "accuracy":         accuracy_score(y_test, y_pred),
    "balanced_accuracy": balanced_accuracy_score(y_test, y_pred),
    "macro_f1":         f1_score(y_test, y_pred, average="macro", zero_division=0),
}
print("\nМетрики на тесте:")
for k, v in test_metrics.items():
    print(f"  {k:18s}: {v:.4f}")

print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))

labels = sorted(pd.unique(y_train_full))
cm = confusion_matrix(y_test, y_pred, labels=labels)
cm_df = pd.DataFrame(cm, index=labels, columns=labels)
print("\nМатрица ошибок (строки — истина, столбцы — предсказание):")
print(cm_df)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(cm_df, annot=True, fmt="d", cmap="Blues", cbar=False, ax=axes[0])
axes[0].set_title(f"Confusion matrix (counts)\n{best_model_name}")
axes[0].set_xlabel("Predicted")
axes[0].set_ylabel("True")

cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)
sns.heatmap(pd.DataFrame(cm_norm, index=labels, columns=labels),
            annot=True, fmt=".2%", cmap="Blues", cbar=False, ax=axes[1])
axes[1].set_title(f"Confusion matrix (row-normalized)\n{best_model_name}")
axes[1].set_xlabel("Predicted")
axes[1].set_ylabel("True")

plt.tight_layout()
plt.show()

# Анализ ошибок: какие классы путаются чаще всего ---
off_diag = cm.copy()
np.fill_diagonal(off_diag, 0)
errors = (
    pd.DataFrame(off_diag, index=labels, columns=labels)
      .stack()
      .reset_index()
)
errors.columns = ["true", "pred", "count"]
errors = errors[errors["count"] > 0].sort_values("count", ascending=False)
print("\nТоп ошибок (true → pred):")
print(errors.head(10).to_string(index=False))
```
